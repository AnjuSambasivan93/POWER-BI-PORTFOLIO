```python
df = spark.table("default.police_victimisations_raw")

display(df)

```

```python
display(spark.sql("SHOW TABLES"))
```

```python
df.printSchema()
print("Total rows", df.count())
```

```python
from pyspark.sql.functions import col, to_date, trim, regexp_replace, year, month
clean_df = (
    df
    .withColumn("Date", to_date(col("Year Month"), "M/d/yyyy"))
    .withColumn(
        "Territorial Authority", 
        trim(regexp_replace(col("Territorial Authority"), r"\.$", "")))
    .withColumn("Year", year("Date"))
    .withColumn("Month", month("Date"))
    .drop("Year Month")
     # Rename columns to remove spaces
    .withColumnRenamed("Age Group", "AgeGroup")
    .withColumnRenamed("Anzsoc Division", "AnzsocDivision")
    .withColumnRenamed("Location Type", "LocationType")
    .withColumnRenamed("Police Area", "PoliceArea")
    .withColumnRenamed("Rov Division", "ROVDivision")
    .withColumnRenamed("Territorial Authority", "TerritorialAuthority")
    .withColumnRenamed("Person/Organisation", "PersonOrganisation")
)
```

```python
clean_df.printSchema()
print("Total rows", clean_df.count())
```

```python
display(clean_df)
```

```python
clean_df.write \
    .format("delta") \
    .mode("overwrite") \
    .partitionBy("Year") \
    .saveAsTable("default.police_victimisations_clean")
```

```python
from pyspark.sql.functions import sum

gold_df = (
    clean_df.groupBy("Date", "Year", "Month", "AgeGroup", "AnzsocDivision", "LocationType", "PoliceArea", "TerritorialAuthority", "ROVDivision", "PersonOrganisation")
    .agg(sum("Victimisations").alias("TotalVictimizations")
    ))
```

```python
display(gold_df)
```

```python
gold_df.write \
    .format("delta") \
    .mode("overwrite") \
    .partitionBy("Year") \
    .saveAsTable("default.police_victimisations_gold")
```
